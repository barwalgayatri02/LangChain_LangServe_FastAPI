```python
from fastapi import FastAPI
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI
from langserve import add_routes
from langchain_community.llms import Ollama
from dotenv import load_dotenv
import os
import nest_asyncio
import uvicorn

# Load .env
load_dotenv()

# Allow nested loops (required for notebooks)
nest_asyncio.apply()

# Set API key
os.environ['OPENAI_API_KEY'] = os.getenv("OPENAI_API_KEY")

# FastAPI App
app = FastAPI(
    title="Langchain Server",
    version="1.0",
    description="A simple API Server"
)

# MODELS
openai_model = ChatOpenAI(model="gpt-4o-mini", temperature=0)
llm_ollama = Ollama(model="llama2")

# ROUTES
add_routes(app, openai_model, path="/openai")

prompt1 = ChatPromptTemplate.from_template("Write me an essay about {topic} with 100 words.")
add_routes(app, prompt1 | openai_model, path="/essay")

prompt2 = ChatPromptTemplate.from_template("Write me a poem about {topic} for a 5-year-old child with 100 words.")
add_routes(app, prompt2 | llm_ollama, path="/poem")

# -----------------------------------
# Run inside JUPYTER NOTEBOOK
# -----------------------------------
from uvicorn import Config, Server

config = Config(app=app, host="0.0.0.0", port=8000, reload=False)
server = Server(config=config)

import asyncio
asyncio.get_event_loop().create_task(server.serve())
```
